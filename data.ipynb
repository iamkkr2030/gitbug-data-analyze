```python
from datasets import load_dataset
import pandas as pd

def create_full_csv():
    # 1. Hugging Face에서 데이터셋 로드
    dataset = load_dataset("av9ash/GitBugs")
    
    # 2. 모든 분할 데이터(train, test, validation 등)를 리스트에 담기
    df_list = []
    for split_name in dataset.keys():
        df = dataset[split_name].to_pandas()
        df_list.append(df)
    
    # 3. 리스트에 담긴 데이터프레임들을 하나로 병합 (데이터 분할 없이 통합)
    full_df = pd.concat(df_list, ignore_index=True)
    
    # 4. 전체 데이터를 하나의 CSV 파일로 저장
    full_df.to_csv("gitbugs_full_data.csv", index=False)
    print(f"전체 데이터가 gitbugs_full_data.csv로 저장되었습니다. (총 데이터 수: {len(full_df)}행)")

if __name__ == "__main__":
    create_full_csv()
```
